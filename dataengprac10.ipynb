```python
# ============================================================
# PRACTICAL - 10
# END-TO-END E-COMMERCE DATA ENGINEERING PIPELINE
# ============================================================

import pandas as pd
import sqlite3
import matplotlib.pyplot as plt
from IPython.display import display


# ============================================================
# 1. DATA INGESTION
# ============================================================

print("========== DATA INGESTION ==========")

# Customers
customers = pd.DataFrame({
    "customer_id": [1, 2, 3, 4, 5],
    "customer_name": ["John", "Mary", "David", "Alice", "Bob"],
    "city": ["Mumbai", "Delhi", "Pune", "Mumbai", "Chennai"]
})

# Products
products = pd.DataFrame({
    "product_id": [101, 102, 103, 104, 105],
    "product_name": [
        "Laptop",
        "Phone",
        "Tablet",
        "Headphones",
        "Keyboard"
    ],
    "category": [
        "Electronics",
        "Electronics",
        "Electronics",
        "Accessories",
        "Accessories"
    ],
    "price": [60000, 30000, 20000, 3000, 1500]
})

# Orders
orders = pd.DataFrame({
    "order_id": [1001, 1002, 1003, 1004, 1005],
    "customer_id": [1, 2, 3, 4, 5],
    "product_id": [101, 102, 103, 104, 105],
    "quantity": [1, 2, 1, 3, 2]
})

# Payments
payments = pd.DataFrame({
    "payment_id": [501, 502, 503, 504, 505],
    "order_id": [1001, 1002, 1003, 1004, 1005],
    "payment_status": [
        "Paid",
        "Paid",
        "Paid",
        "Pending",
        "Paid"
    ]
})


print("\nCustomers:")
display(customers)

print("\nProducts:")
display(products)

print("\nOrders:")
display(orders)

print("\nPayments:")
display(payments)


# ============================================================
# 2. SAVE DATA AS CSV FILES
# ============================================================

print("\n========== CSV FILE CREATION ==========")

customers.to_csv("customers.csv", index=False)
products.to_csv("products.csv", index=False)
orders.to_csv("orders.csv", index=False)
payments.to_csv("payments.csv", index=False)

print("All CSV files created successfully!")


# ============================================================
# 3. DATA CLEANING
# ============================================================

print("\n========== DATA CLEANING ==========")

# Check missing values

print("\nMissing values in Customers:")
print(customers.isnull().sum())

print("\nMissing values in Products:")
print(products.isnull().sum())

print("\nMissing values in Orders:")
print(orders.isnull().sum())

print("\nMissing values in Payments:")
print(payments.isnull().sum())


# Remove duplicate records

customers = customers.drop_duplicates()
products = products.drop_duplicates()
orders = orders.drop_duplicates()
payments = payments.drop_duplicates()

print("\nDuplicates removed successfully!")


# Check data types

print("\nCustomers Data Types:")
print(customers.dtypes)

print("\nProducts Data Types:")
print(products.dtypes)

print("\nOrders Data Types:")
print(orders.dtypes)

print("\nPayments Data Types:")
print(payments.dtypes)


# Display cleaned datasets

print("\nCleaned Customers:")
display(customers)

print("\nCleaned Products:")
display(products)

print("\nCleaned Orders:")
display(orders)

print("\nCleaned Payments:")
display(payments)


# ============================================================
# 4. DATA TRANSFORMATION
# ============================================================

print("\n========== DATA TRANSFORMATION ==========")


# Join Orders and Products

order_data = orders.merge(
    products,
    on="product_id",
    how="inner"
)

print("\nOrders + Products:")
display(order_data)


# Calculate total sales

order_data["total_sales"] = (
    order_data["price"] * order_data["quantity"]
)

print("\nTotal Sales Calculated:")
display(order_data)


# Add customer information

order_data = order_data.merge(
    customers,
    on="customer_id",
    how="inner"
)

print("\nAdded Customer Information:")
display(order_data)


# Add payment information

order_data = order_data.merge(
    payments,
    on="order_id",
    how="inner"
)

print("\nFinal Transformed Dataset:")
display(order_data)


# ============================================================
# 5. SALES SUMMARY
# ============================================================

print("\n========== SALES SUMMARY ==========")

sales_summary = order_data.groupby("category").agg(
    total_sales=("total_sales", "sum"),
    total_quantity=("quantity", "sum")
).reset_index()

print("\nSales Summary by Category:")
display(sales_summary)


# ============================================================
# 6. CREATE SQL DATABASE
# ============================================================

print("\n========== SQL DATABASE STORAGE ==========")

connection = sqlite3.connect("ecommerce_warehouse.db")

print("Database created successfully!")


# ============================================================
# 7. STORE TABLES IN DATABASE
# ============================================================

customers.to_sql(
    "dim_customer",
    connection,
    if_exists="replace",
    index=False
)

products.to_sql(
    "dim_product",
    connection,
    if_exists="replace",
    index=False
)

orders.to_sql(
    "orders",
    connection,
    if_exists="replace",
    index=False
)

payments.to_sql(
    "payments",
    connection,
    if_exists="replace",
    index=False
)

order_data.to_sql(
    "fact_sales",
    connection,
    if_exists="replace",
    index=False
)

print("All tables stored in the database!")


# ============================================================
# 8. DISPLAY DATABASE TABLES
# ============================================================

print("\n========== DATABASE TABLES ==========")

print("\nCustomer Dimension:")
display(
    pd.read_sql_query(
        "SELECT * FROM dim_customer",
        connection
    )
)

print("\nProduct Dimension:")
display(
    pd.read_sql_query(
        "SELECT * FROM dim_product",
        connection
    )
)

print("\nFact Sales:")
display(
    pd.read_sql_query(
        "SELECT * FROM fact_sales",
        connection
    )
)


# ============================================================
# 9. CHECK DATABASE TABLES
# ============================================================

tables = pd.read_sql_query(
    """
    SELECT name
    FROM sqlite_master
    WHERE type='table'
    """,
    connection
)

print("\nTables in Data Warehouse:")
display(tables)


# ============================================================
# 10. DATE DIMENSION
# ============================================================

print("\n========== DATE DIMENSION ==========")

date_data = pd.DataFrame({
    "date_id": [1, 2, 3, 4, 5],

    "date": [
        "2026-01-01",
        "2026-01-02",
        "2026-01-03",
        "2026-01-04",
        "2026-01-05"
    ],

    "year": [
        2026,
        2026,
        2026,
        2026,
        2026
    ],

    "month": [
        1,
        1,
        1,
        1,
        1
    ]
})

print("\nDate Dimension:")
display(date_data)

date_data.to_sql(
    "dim_date",
    connection,
    if_exists="replace",
    index=False
)

print("Date dimension created successfully!")


# ============================================================
# 11. REPORTING - TOTAL SALES
# ============================================================

print("\n========== REPORTING ==========")

query = """
SELECT SUM(total_sales) AS total_sales
FROM fact_sales
"""

total_sales_report = pd.read_sql_query(
    query,
    connection
)

print("\nTotal Sales:")
display(total_sales_report)


# ============================================================
# 12. SALES BY CATEGORY
# ============================================================

query = """
SELECT
    category,
    SUM(total_sales) AS total_sales
FROM fact_sales
GROUP BY category
"""

category_report = pd.read_sql_query(
    query,
    connection
)

print("\nSales by Category:")
display(category_report)


# ============================================================
# 13. SALES BY CUSTOMER
# ============================================================

query = """
SELECT
    customer_name,
    SUM(total_sales) AS total_sales
FROM fact_sales
GROUP BY customer_name
"""

customer_report = pd.read_sql_query(
    query,
    connection
)

print("\nSales by Customer:")
display(customer_report)


# ============================================================
# 14. PAYMENT REPORT
# ============================================================

query = """
SELECT
    payment_status,
    COUNT(*) AS number_of_orders,
    SUM(total_sales) AS total_sales
FROM fact_sales
GROUP BY payment_status
"""

payment_report = pd.read_sql_query(
    query,
    connection
)

print("\nPayment Report:")
display(payment_report)


# ============================================================
# 15. DATA VISUALIZATION
# ============================================================

print("\n========== DATA VISUALIZATION ==========")

plt.figure(figsize=(8, 5))

plt.bar(
    category_report["category"],
    category_report["total_sales"],
    color=["steelblue", "orange"]
)

plt.title("Sales by Product Category")
plt.xlabel("Category")
plt.ylabel("Total Sales")

plt.tight_layout()
plt.show()


# ============================================================
# 16. SALES BY CUSTOMER VISUALIZATION
# ============================================================

plt.figure(figsize=(8, 5))

plt.bar(
    customer_report["customer_name"],
    customer_report["total_sales"],
    color="green"
)

plt.title("Sales by Customer")
plt.xlabel("Customer")
plt.ylabel("Total Sales")

plt.tight_layout()
plt.show()


# ============================================================
# 17. PAYMENT STATUS VISUALIZATION
# ============================================================

plt.figure(figsize=(7, 5))

plt.bar(
    payment_report["payment_status"],
    payment_report["number_of_orders"],
    color=["green", "red"]
)

plt.title("Orders by Payment Status")
plt.xlabel("Payment Status")
plt.ylabel("Number of Orders")

plt.tight_layout()
plt.show()


# ============================================================
# 18. FINAL PROJECT STATUS
# ============================================================

print("\n============================================")
print("END-TO-END DATA ENGINEERING PROJECT")
print("============================================")

print("✓ Data Ingestion")
print("✓ Data Cleaning")
print("✓ Data Transformation")
print("✓ CSV File Creation")
print("✓ SQL Database Storage")
print("✓ Data Warehouse Design")
print("✓ Date Dimension")
print("✓ Sales Reporting")
print("✓ Customer Reporting")
print("✓ Payment Reporting")
print("✓ Data Visualization")

print("\nPROJECT COMPLETED SUCCESSFULLY!")


# ============================================================
# 19. CLOSE DATABASE CONNECTION
# ============================================================

connection.close()

print("\nSQLite database connection closed.")
```
