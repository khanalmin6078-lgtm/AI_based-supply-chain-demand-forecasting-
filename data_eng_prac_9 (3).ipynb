1. CSV → Pandas → Data Cleaning → SQL Database → Report

```python
import pandas as pd

data = {
    "Customer": ["John", "Mary", "John", "David", None],
    "Product": ["Laptop", "Phone", "Laptop", "Tablet", "Phone"],
    "Sales": [1000, 500, 1000, 700, None],
    "Quantity": [1, 2, 1, 3, 2]
}

df = pd.DataFrame(data)

df.to_csv("sales_data.csv", index=False)

print("CSV created successfully!")
df
```

```python
df = pd.read_csv("sales_data.csv")

print("Original Data:")
print(df)
```

```python
print("Missing values:")
print(df.isnull().sum())
```

```python
df["Customer"] = df["Customer"].fillna("Unknown")
df["Sales"] = df["Sales"].fillna(df["Sales"].mean())

print("Cleaned Data:")
print(df)
```

```python
df = df.drop_duplicates()

print("After removing duplicates:")
print(df)
```

```python
import sqlite3

connection = sqlite3.connect("sales_database.db")

df.to_sql(
    "sales",
    connection,
    if_exists="replace",
    index=False
)

print("Data stored in SQL database successfully!")
```

```python
query = """
SELECT Product, SUM(Sales) AS Total_Sales
FROM sales
GROUP BY Product
"""

report = pd.read_sql_query(query, connection)

print("Sales Report:")
print(report)
```

```python
connection.close()

print("Pipeline completed successfully!")
```

2. ETL pipeline for an e-commerce dataset

```python
import pandas as pd

customers = pd.DataFrame({
    "customer_id": [1, 2, 3],
    "name": ["John", "Mary", "David"],
    "email": ["john@gmail.com", "mary@gmail.com", "david@gmail.com"]
})

products = pd.DataFrame({
    "product_id": [101, 102, 103],
    "product": ["Laptop", "Phone", "Tablet"],
    "price": [1000, 500, 700]
})

orders = pd.DataFrame({
    "order_id": [1001, 1002, 1003],
    "customer_id": [1, 2, 3],
    "product_id": [101, 102, 103],
    "quantity": [1, 2, 1]
})

payments = pd.DataFrame({
    "payment_id": [501, 502, 503],
    "order_id": [1001, 1002, 1003],
    "payment_status": ["Paid", "Paid", "Pending"]
})

print("Data extracted successfully!")
```

```python
order_data = orders.merge(
    products,
    on="product_id"
)

order_data["total_amount"] = (
    order_data["price"] * order_data["quantity"]
)

print("Transformed Order Data:")
print(order_data)
```

```python
order_data = order_data.merge(
    customers,
    on="customer_id"
)

print("Customer + Order Data:")
print(order_data)
```

```python
order_data = order_data.merge(
    payments,
    on="order_id"
)

print("Complete E-commerce Data:")
print(order_data)
```

```python
import sqlite3

connection = sqlite3.connect("ecommerce.db")

customers.to_sql(
    "customers",
    connection,
    if_exists="replace",
    index=False
)

products.to_sql(
    "products",
    connection,
    if_exists="replace",
    index=False
)

orders.to_sql(
    "orders",
    connection,
    if_exists="replace",
    index=False
)

payments.to_sql(
    "payments",
    connection,
    if_exists="replace",
    index=False
)

order_data.to_sql(
    "order_report",
    connection,
    if_exists="replace",
    index=False
)

print("ETL pipeline completed!")
```

```python
query = """
SELECT product,
       SUM(quantity) AS total_quantity,
       SUM(total_amount) AS total_sales
FROM order_report
GROUP BY product
"""

report = pd.read_sql_query(query, connection)

print("E-commerce Sales Report:")
print(report)
```

```python
connection.close()
```

3. Historical + newly arriving data pipeline

```python
import pandas as pd

historical_data = pd.DataFrame({
    "order_id": [1, 2, 3],
    "customer": ["John", "Mary", "David"],
    "sales": [1000, 500, 700]
})

historical_data
```

```python
new_data = pd.DataFrame({
    "order_id": [4, 5],
    "customer": ["Alice", "Bob"],
    "sales": [800, 600]
})

new_data
```

```python
combined_data = pd.concat(
    [historical_data, new_data],
    ignore_index=True
)

print("Combined Data:")
print(combined_data)
```

```python
combined_data = combined_data.drop_duplicates(
    subset=["order_id"]
)

print("Data after removing duplicates:")
print(combined_data)
```

```python
import sqlite3

connection = sqlite3.connect("historical_new_data.db")

combined_data.to_sql(
    "orders",
    connection,
    if_exists="replace",
    index=False
)

print("Historical and new data loaded into database!")
```

```python
query = """
SELECT customer,
       SUM(sales) AS total_sales
FROM orders
GROUP BY customer
"""

final_report = pd.read_sql_query(
    query,
    connection
)

print("Final Report:")
print(final_report)
```

```python
connection.close()

print("Historical + New Data Pipeline Completed!")
```
